# MAYFLIES POPULATION MODEL 

Population models are interesting in mathematics as they try to predict the changes in the population over a certain period. These are not restricted to humans, as it can help to find out how many numbers of cells are going to be in a cell culture lab in a certain amount of time. But those type of calculation take complex functions. This project intends to analysze the behavior of a simple population module, which is mayflies. These are insects that spend at most a few years on the bottom of the river They then emerge from the water over several days, mate, lay eggs and die Therefore, only one generation of mayflies are alive at any given time. The population module of mayfly, called *Mayflies Model* is as follows: $$ y_n = b(1-y_{n-1}) y_{n-1} $$ where b=aM and the range of b is between 0 and 4. In here, M= the max population and a is the the range of 1 and 4/M. $y_n$ is the population of mayflies in that year (n representing the number of years). Which implies that $y_{n-1}$ is the population present in the previous year. [1]

This project tries to analyze the mayflies’ model for various values of 𝑏 and $𝑦_0$. To do this, there are 3 user defined functions used, namely: a) basic_population(t,initial_population,b) b)Mayflies_model(t,initial_population,width) and c)Mayflies_model_b(t,b,width). The user defined functions are explained with the codes down below. You wiil find a common function being used everywhere named basic_population(t,initial_population,b), which just makes the population to be used.

matplotlib.pyplot is imported as the python library as it creates the graph.

```python
import matplotlib.pyplot as plt # imports a python library that gives the graph. 
```

This function named basic_population(t,initial_population,b) gives the population to be used in the population module. The formula used to calculate this population is $$ y_n = b(1-y_{n-1}) y_{n-1} $$

```python
def basic_population(t,initial_population,b): 
    population=[initial_population] # the first element (zeroth index) of population is the initial population.
    for i in range (t): # variable i goes through t 
        population.append(b*(1-population[-1])*population[-1]) # multiplies b to the previous value of population, which is multiplied to the value obtained when 1 is subtracted from the previous value of population. This value is added to population.
    return population #returns the list population.
```

This function named Mayflies_model(t,initial_population,width) is a function which will give a graph with x -axis as the b value and y-axis as the final population. The initial population will be inputed by the user. The user is recommended to input a list of initial population as to see the difference between different values of the initial population. From [2], this is also call a bifurcation of b as the output of the model changes when **one** of the parameter changes, which in this case is **initial population**.

**The values of initial population MUST be between 0 and 1** [1]

```python
#Bifurcation of B
initial_population=[] #declaring initial population to be a list
def Mayflies_model(t,initial_population,width):
    b=[i/10000 for i in range (0,40000)] #the value i goes from 0 to 40000 and return i/10000. This created many plot points and makes the curve a lot more smoother.
    plt.figure(figsize=[width,width*0.5]) # specifies the dimensions of the graph.
    plt.xlabel('b values')
    plt.ylabel('final population')
    plt.title('#Bifurcation of b ')
    for j in range(len(initial_population)): # j goes through the initial value inputed by the user
        final_population= [basic_population(t,initial_population[j],i)[-1] for i in b] # goes through the function: basic population for every value of b and adds the last element of the list 'population' value to final population
        plt.plot(b,final_population,label='initial population is {}'.format(initial_population[j])) #plots the graph with x as b calues and y as final population values. Also gives labels to the curves.
    plt.legend() # assigning meaning to the various plot elements
```

Function call for Mayflies_model. The ideal time period would be between 5-10 years. 

```python
Mayflies_model(100,[0.1,0.2],30) #input the time, initial population and the width of the graph.
```

This function named Mayflies_model_b(t,b,width) is a function which will give a graph with x -axis as the initial population and y-axis as the final population. The b values will be inputed by the user. The user is recommended to input a list of b values as to see the difference between different values of the b. From [2], this is also call a bifurcation of initial population as the output of the model changes when **one** of the parameter changes, which in this case is the b value.

**The values of b MUST be between 0 and 4** [1]

```python
def Mayflies_model_b(t,b,width): 
    initial_population= [i/1000 for i in range (0,1000)] #the value i goes from 0 to 1000 and return i/1000. This created many plot points and makes the curve a lot more smoother.
    plt.figure(figsize=[width,width*0.5]) # specifies the dimensions of the graph. 
    plt.xlabel('initial_population')
    plt.ylabel('final population')
    plt.title('bifurcation of initial population ')
    for j in range(len(b)):  #j goes through the b inputed by the user
        final_population= [basic_population(t,i,b[j])[-1] for i in initial_population] # goes through the function: basic population for every value of initial_population and adds last element of the list 'population' value to final population
        plt.plot(initial_population,final_population,label='b is {}'.format(b[j])) #plots the graph with x as initial population calues and y as final population values. Also gives labels to the curves.
    plt.legend() # assigning meaning to the various plot elements    plt.legend()  
    plt.show()
```

Function call for Mayflies_model_b. The ideal time period would be between 5-10 years.

```python
Mayflies_model_b(100,[2.7,3,3.2,3.5],30) #input the time, b-values and the width of the graph.
```
